## 5. Train/Test Split

```python
data['date'] = pd.to_datetime(data['date'])
data['year'] = data['date'].dt.year
data['month'] = data['date'].dt.month

labels = data['price']
train1 = data.drop(['id', 'price', 'date'], axis=1)
train1.head()
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    train1, labels, test_size=0.10, random_state=2
)
print("Train size:", x_train.shape, " Test size:", x_test.shape)
```

## 6. **Compare** **Multiple** **Algorithms**

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=1.0),
    "Decision Tree": DecisionTreeRegressor(max_depth=8, random_state=2),
    "Random Forest": RandomForestRegressor(n_estimators=300, max_depth=12, random_state=2, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=400, max_depth=5, min_samples_split=2,
        learning_rate=0.1, loss='squared_error', random_state=2
    ),
}

trained_models = {}
scores = {}

for name, model in models.items():
    model.fit(x_train, y_train)
    acc = model.score(x_test, y_test)
    trained_models[name] = model
    scores[name] = acc
    print(f"{name:20s} -> R^2 Accuracy: {acc:.4f}  ({acc*100:.2f}%)")
```

## 7. Accuracy Comparison Chart

```python
results = pd.DataFrame({
    'Model': list(scores.keys()),
    'R^2 Accuracy': list(scores.values())
}).sort_values('R^2 Accuracy', ascending=False).reset_index(drop=True)

results['Accuracy %'] = (results['R^2 Accuracy'] * 100).round(2)
results
```

```python
plt.figure(figsize=(9,5))
plt.barh(results['Model'], results['Accuracy %'])
plt.xlabel('Accuracy (%)')
plt.title('Algorithm Accuracy Comparison — House Price Prediction')
plt.gca().invert_yaxis()
for i, v in enumerate(results['Accuracy %']):
    plt.text(v + 0.5, i, f"{v}%", va='center')
plt.show()
```

## 8. Best Model — Final Selection

```python
best_model_name = results.loc[0, 'Model']
best_score = results.loc[0, 'R^2 Accuracy']
clf = trained_models[best_model_name]   # final chosen model

print(f"BEST ALGORITHM: {best_model_name}")
print(f"FINAL ACCURACY (R^2): {best_score:.4f}  ({best_score*100:.2f}%)")
```

## 9. Training vs Test Deviance Curve (Gradient Boosting)

```python
gb_model = trained_models["Gradient Boosting"]
params = {'n_estimators': 400}

test_score = np.zeros((params['n_estimators'],), dtype=np.float64)

for i, y_pred in enumerate(gb_model.staged_predict(x_test)):
    test_score[i] = np.mean((y_test - y_pred) ** 2)

estimators_range = np.arange(params['n_estimators']) + 1

plt.figure(figsize=(10, 6))
plt.plot(estimators_range, gb_model.train_score_, 'b-', label='Training Set Deviance')
plt.plot(estimators_range, test_score, 'r-', label='Test Set Deviance')
plt.legend(loc='upper right')
plt.xlabel('Boosting Iterations')
plt.ylabel('Deviance (MSE)')
plt.title('Gradient Boosting: Training vs Test Deviance')
plt.show()
```

```python
if hasattr(clf, "feature_importances_"):
    importances = pd.Series(clf.feature_importances_, index=train1.columns).sort_values(ascending=False)

    plt.figure(figsize=(10, 6))
    importances.plot(kind='bar')
    plt.title(f"Feature Importance — {best_model_name}")
    plt.ylabel("Importance")
    plt.show()

    print(importances)
else:
    print(f"{best_model_name} does not expose feature_importances_ (it's a linear model).")
```
